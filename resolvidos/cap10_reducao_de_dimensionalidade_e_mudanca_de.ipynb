### Aplicação: compressão de imagens · PDF p. 212

#### Exercício: Compressão de imagem com SVD

Crie uma "imagem" 100×100 (por ex., `outer(sin(1:100/8), cos(1:100/12))` + um quadrado + ruído). Reconstrua com os 1, 5 e 20 primeiros valores singulares e mostre lado a lado com `image()`.

```r
img <- outer(sin(1:100 / 8), cos(1:100 / 12)); img[30:60, 40:70] <- img[30:60, 40:70] + 1
img <- img + matrix(rnorm(1e4, sd = 0.1), 100)
s <- svd(img)
rec <- function(k) s$u[, 1:k, drop = FALSE] %*% diag(s$d[1:k], k) %*% t(s$v[, 1:k, drop = FALSE])
par(mfrow = c(1, 4), mar = c(1, 1, 2, 1))
image(img, main = "original", axes = FALSE)
for (k in c(1, 5, 20)) image(rec(k), main = paste(k, "componentes"), axes = FALSE)
par(mfrow = c(1, 1))
```

## Projeções Aleatórias · PDF p. 219

#### Exercício: Projeções aleatórias preservam distâncias

Gere 50 pontos em 1.000 dimensões e projete em 100 dimensões com uma matriz normal / √100. Compare as distâncias entre pares antes e depois (razão média e desvio).

```r
X <- matrix(rnorm(50 * 1000), 50)
R <- matrix(rnorm(1000 * 100), 1000) / sqrt(100)
razao <- as.vector(dist(X %*% R)) / as.vector(dist(X))
c(media = mean(razao), desvio = sd(razao))
```
