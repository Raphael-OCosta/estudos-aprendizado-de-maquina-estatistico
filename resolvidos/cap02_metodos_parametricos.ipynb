## O Método dos Mínimos Quadrados · PDF p. 46

#### Exercício: Mínimos quadrados com matrizes

Para `mpg ~ wt + hp` (base `mtcars`), calcule β = (XᵀX)⁻¹Xᵀy com `solve` e `model.matrix`, e confira com `coef(lm(...))`.

```r
X <- model.matrix(mpg ~ wt + hp, mtcars); y <- mtcars$mpg
beta <- solve(t(X) %*% X, t(X) %*% y)
cbind(na_mao = drop(beta), lm = coef(lm(mpg ~ wt + hp, mtcars)))
```

### Mínimos Quadrados quando a Suposição de Linearidade Falha · PDF p. 47

#### Exercício: Quando a reta não basta

Nos dados do seno, compare o EQM de teste de `y ~ x` com `y ~ poly(x, 5)`. Por que mínimos quadrados ainda serve, mesmo a relação não sendo linear em x?

```r
sim_seno <- function(n, sd = 0.3) {
  x <- runif(n)
  data.frame(x = x, y = sin(2 * pi * x) + rnorm(n, sd = sd))
}
treino <- sim_seno(150); teste <- sim_seno(2000)
eqm <- function(m) mean((teste$y - predict(m, teste))^2)
c(reta = eqm(lm(y ~ x, treino)), poli5 = eqm(lm(y ~ poly(x, 5), treino)))
# O modelo continua linear nos parâmetros: só trocamos as covariáveis (x, x², ..., x⁵).
```

### Regressão Linear no R · PDF p. 48

#### Exercício: lm, summary e predict

Com `airquality`, ajuste `Ozone ~ Temp + Wind`. Mostre o `summary` e preveja o ozônio para Temp = 85 e Wind = 7, com intervalo de predição de 95%.

```r
m <- lm(Ozone ~ Temp + Wind, data = airquality)
summary(m)
predict(m, data.frame(Temp = 85, Wind = 7), interval = "prediction")
```
