## K-Médias · PDF p. 228

#### 🧪 Exercício: K-médias e o cotovelo

Rode `kmeans` nas 4 medidas da `iris` (padronizadas) para k = 1 a 8 e desenhe a soma de quadrados dentro dos grupos. Depois, com k = 3, compare os grupos com as espécies.

```r
X <- scale(iris[, 1:4])
wss <- sapply(1:8, function(k) kmeans(X, k, nstart = 20)$tot.withinss)
plot(1:8, wss, type = "b", xlab = "k")
table(grupo = kmeans(X, 3, nstart = 20)$cluster, especie = iris$Species)
```

## Métodos Hierárquicos · PDF p. 237

#### 🧪 Exercício: Agrupamento hierárquico

Faça `hclust` em `dist(scale(USArrests))` com ligação completa, desenhe o dendrograma e corte em 4 grupos. Quais estados ficaram no menor grupo?

```r
h <- hclust(dist(scale(USArrests)), method = "complete")
plot(h, cex = 0.6); rect.hclust(h, k = 4)
g <- cutree(h, 4); names(g)[g == names(which.min(table(g)))]
```
