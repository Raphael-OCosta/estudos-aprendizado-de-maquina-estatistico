## Função de Risco · PDF p. 152

#### 🧪 Exercício: Risco 0-1 de uma regra simples

Em `iris`, fique só com versicolor e virginica. Use a regra "virginica se `Petal.Length > 4,8`" e calcule a proporção de erros (risco com perda 0-1).

```r
d <- droplevels(subset(iris, Species != "setosa"))
prev <- ifelse(d$Petal.Length > 4.8, "virginica", "versicolor")
mean(prev != d$Species)
```

## Outras medidas de desempenho · PDF p. 156

#### 🧪 Exercício: Matriz de confusão e métricas

Para a mesma regra, monte a matriz de confusão e calcule sensibilidade, especificidade e precisão, considerando virginica como a classe positiva.

```r
d <- droplevels(subset(iris, Species != "setosa"))
prev <- factor(ifelse(d$Petal.Length > 4.8, "virginica", "versicolor"), levels(d$Species))
tab <- table(previsto = prev, real = d$Species); tab
VP <- tab["virginica", "virginica"]; FN <- tab["versicolor", "virginica"]
FP <- tab["virginica", "versicolor"]; VN <- tab["versicolor", "versicolor"]
c(sensibilidade = VP / (VP + FN), especificidade = VN / (VN + FP), precisao = VP / (VP + FP))
```
